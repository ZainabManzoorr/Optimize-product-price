```python
import joblib
import pandas as pd
```

```python
xgb = joblib.load("xgboost_model.pkl")
feature_names = joblib.load("feature_columns.pkl")
```

```python
xgb_importance = pd.Series(xgb.feature_importances_, index=feature_names)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
xgb_top = xgb_importance.sort_values(ascending=False).head(10)

plt.figure(figsize=(8,6))
sns.barplot(x=xgb_top.values, y=xgb_top.index, palette="magma",hue=xgb_top.values)
plt.title("Top 10 Features (XGBoost)")
plt.xlabel("Importance Score")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```

```python
xgb_importance.to_csv("xgb_feature_importance.csv", header=["importance"])
```
